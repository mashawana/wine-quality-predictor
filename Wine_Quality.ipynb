```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

#loading the data
red_wine = pd.read_csv(r"C:\Users\Mpho\Downloads\wine+quality\winequality-red.csv", sep= ";")
red_wine.head()

white_wine = pd.read_csv(r"C:\Users\Mpho\Downloads\wine+quality\winequality-white.csv", sep= ";")
white_wine.head()

#adding new columns to identify the wine type before combining the two datasets
red_wine["wine_type"] = "red"
white_wine["wine_type"] = "white"

#combining the two datasets
wine_quality = pd.concat([red_wine, white_wine], ignore_index=True)
print(wine_quality.head())
```

```python
# Exploratory Data Analysis (EDA)
# checking for missing values
print(wine_quality.info())
print(f"\nMissing values per column: {wine_quality.isnull().sum().sum()}")

#finding which wine type has the highest quality rating
#we have to drop tjhe wine_type column to find the highest quality rating
correlation = wine_quality.drop(columns=["wine_type"]).corr()["quality"].sort_values(ascending=False)
print(f"\nCorrelation with quality:\n{correlation}")

#creating binary target: 1 for high quality(Quality >=7) and 0 for low quality (Quality <7)
wine_quality["is_good"] = (wine_quality["quality"] >= 7).astype(int)

#checking the ditribution for the new binary target variable
print(f"\nClass distribution (0 = Average, 1 = Good):")
print(wine_quality["is_good"].value_counts(normalize=True))
```

```python
# Data processing and splitting

#separating features and target variable
X = wine_quality.drop(columns=["quality", "is_good"])
Y = wine_quality["is_good"]

#encoding wine_type column into numbers (red = 0, white = 1)
X["wine_type"] = X["wine_type"].map({"red": 0, "white": 1})

#splitting the dataset into training and testing sets
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 42, stratify= Y)

#scaling the data using StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"\nTraining set shape: {X_train_scaled.shape}")
print(f"Testing set shape: {X_test_scaled.shape}")
```
